## Clean Data
The raw dataset has a lot of un-usable reviews, so we must clean it before we can continue with the analysis.

These un-usable rows have at least one of the following in the "review column:
- NULL value
- Empty string review
- contains only a number, date, or special characters
- non-english text

We have also removed all whitespace from around the actual review text (this includes starting with newlines and tabs).

```python
# Removes rows that have NULL reviews
BG3_df.dropna(inplace=True)

# Remove rows where "review" is an empty string
BG3_df = BG3_df[BG3_df["review"] != ""]

# Remove rows where "review" is a number, date, or only special characters (with no words)
BG3_df = BG3_df[~BG3_df["review"].str.match(r'^\d+|\d{4}-\d{2}-\d{2}$|^[^\w\s]+$', na=False)]

# Remove rows with non-English text in the "review" column
BG3_df = BG3_df[BG3_df["review"].apply(lambda x: bool(re.match(r'^[a-zA-Z0-9\s\.,!?\'"]+$', x)))]

# Strip trailing whitespace from all rows in the "review" column
BG3_df["review"] = BG3_df["review"].str.strip()
```
